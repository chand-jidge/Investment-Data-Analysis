# Business Insights

```python
# Gender Distribution
df["gender"].value_counts()
```

```python
df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.ylabel("")
plt.show()
```

```python
# Age Distribution
df['age'].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=8, kde=True)
plt.title("Age Distribution")
plt.show()

```

```python
# Most Preferred Investment Avenue

investment = ['Mutual_Funds', 'Equity_Market', 'Fixed_Deposits', 'PPF', 'Gold']
df[investment].sum().sort_values(ascending=False)
```

```python
# Bar Chart
df[investment].sum().sort_values().plot(kind='barh')
plt.title('Investment Preferences')
plt.show()
```

```python
# Investment Objective
df['Objective'].value_counts
```

```python
# column chart
sns.countplot(data=df, x='Objective')
plt.xticks(rotation=30)
plt.show()
```
